```python
from glob import glob
from natsort import natsorted
from BudaOCR.Evaluation import Evaluator
from BudaOCR.Utils import read_ctc_model_config, get_kenlm_config, save_json
```

```python
# KenLM Wylie models
kenlm_wylie_5gram = "Models/KenLM/wylie_5gram/wylie_5gram.binary"
kenlm_wlyie_5gram_arpa = "Models/KenLM/wylie_5gram/wylie_5gram.arpa"

kenlm_wylie_6gram = "Models/KenLM/wylie_6gram/wylie_6gram.binary"
kenlm_wlyie_6gram_arpa = "Models/KenLM/wylie_6gram/wylie_6gram.arpa"

kenlm_wylie_config_5g = get_kenlm_config(kenlm_wylie_5gram, kenlm_wlyie_5gram_arpa)
kenlm_wylie_config_6g = get_kenlm_config(kenlm_wylie_6gram, kenlm_wlyie_6gram_arpa)

# KenLM stack models
kenlm_stack_5gram = "Models/KenLM/stack_5gram/stack_5gram.binary"
kenlm_stack_5gram_arpa = "Models/KenLM/stack_5gram/stack_5gram.arpa"

kenlm_stack_config_5g = get_kenlm_config(kenlm_stack_5gram, kenlm_stack_5gram_arpa)
```

#### Easter2-Vit Models

```python
#vit_model_config_file = "Models/2026_2_5_17_23_Easter2-Vit_Wylie/model_config.json"
vit_model_config_file = "Models/2026_2_4_22_51_Easter2-Vit_Stacks/model_config.json"
vit_model_config = read_ctc_model_config(vit_model_config_file)
```

```python
# data
data_root = "/Users/eric/Desktop/Data/Evaluation/LhasaKanjur"

images = natsorted(glob(f"{data_root}/images/*.jpg"))
labels = natsorted(glob(f"{data_root}/labels/*.txt"))

print(f"Images: {len(images)} - Labels: {len(labels)}")
```

```python
vit_model_config_file = "Models/2026_2_4_22_51_Easter2-Vit_Stacks/model_config.json"
vit_model_config = read_ctc_model_config(vit_model_config_file)

evaluator = Evaluator(vit_model_config, device="mps", kenlm_config=kenlm_stack_config_5g)
results = evaluator.evaluate(images, labels)

results_out = f"{data_root}/easter2_vit_stacks_kenlm5g.json"
save_json(results_out, results)
```

```python
model_config_file = "Models/2026_2_4_22_51_Easter2-Vit_Stacks/model_config.json"
model_config = read_ctc_model_config(model_config_file)

evaluator = Evaluator(model_config, device="mps")
results = evaluator.evaluate(images, labels)

results_out = f"{data_root}/easter2_vit_stacks.json"
save_json(results_out, results)
```
